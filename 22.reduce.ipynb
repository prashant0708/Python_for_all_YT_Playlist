## Reducer

Python reduce() is tool from functools modules what applies the given function cumulatively to item of iterable  and return single value

reduce() takes many values and reduces them to ONE final value.



map() → many → many

filter() → many → fewer

reduce() → many → ONE


Python reduce() function is a toll from the functool , 

we import from functool to use it.

```python
## syntax
reduce(function, iterable,initializer)

```

```python
from functools import reduce
nums = [1,2,3,4,5,6,7]

## find the total sum for the given list 


result = reduce(lambda x ,y: x+y, nums)

print(result)
```

```python
result = 0
for i in nums:
    result+=i
print(result)
    
```

```python
sales = {
    "Jan": 50000,
    "Feb": 60000,
    "Mar": 55000
}

## total sales

total_sales = reduce(lambda a,b : a+b , sales.values())

print(total_sales)
```

```python
names = {
    "u1": "Prashant",
    "u2": "Rahul",
    "u3": "Amit"
}

## total count of words for all the names

total_words = reduce(lambda a,b:  a+len(b)  , names.values(),0)
print(total_words)
```

```python
total_words = reduce(lambda a,b:  len(a)+len(b)  , names.values())
print(total_words)
```

```python
## find the students with highest marks

students = {
    "Amit": 78,
    "Neha": 92,
    "Rahul": 85
}

topper = reduce(lambda a,b :   a if a[1]>b[1] else b ,students.items())

print(topper)
```

```python
## calculate total amount and quantity

orders = {
    "order1": 500,
    "order2": 1500,
    "order3": 1000
}

summary = reduce(lambda x ,y :{"total":x['total']+y[1],
                              "count":x["count"]+1
                              
                              
                              },
                 orders.items(), 
                 {"total":0,"count":0}
                 
                 
                 
                 )
```
