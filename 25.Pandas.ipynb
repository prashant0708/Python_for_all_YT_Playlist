```python
## function that will take n number of list to create dataframe out of that
import logging

logging.basicConfig(level=logging.INFO)

def list_to_dataframe(*lists): ## 3 list of 5 length  ### case 2 --> 3 list 5,4,5
    try:
        length_iterator = list(map(lambda x : len(x),lists)) # [5,5,5] , ##[5,4,5] -->[5,4]
        
        logging.info(f"length of each list  is length_iterator = {length_iterator}")
        
        x = set(length_iterator) # [4,4,4] --> {4}
        logging.info(f"length of x after removing the duplicate x= {x}")
        
        if len(x)!=1:
            raise ValueError("length of the given list is not same")
        else:
            data=list(zip(*lists))
            logging.info(f"here is your data after zipping{data}")
            result = pd.DataFrame(data)
        return result
    except Exception as e:
        logging.error({"type":"error", "info":str(e)})  
```

risk_factor 

if age>50 and Policy type == "Basic"  ==> High risk
if age>50 and Policy type =="Gold" ==>  Normal risk
if age>50 and Policy type == "Platinum" ==> Normal

```python
def risk_factor(row):
    if row["Patient_Age"] >= 50 and row["Policy_Type"] == "Basic":
        return "High Risk"
    elif row['Patient_Age']>=50 and row["Policy_Type"]=="Gold":
        return "Normal Risk"
    elif row['Patient_Age']>=50 and row['Policy_Type'] =='Platinum':
        return "Normal"
    else:
        return "Age is below 50"
```

## Pivot and Pivot tables

Both pivot and pivot tables convert the data , reshape the data (row converts into columns)

`syntax of pivot `

- pivot() is used to reshape data but it cannot handle duplicate values.

## Resampling in the pandas

resampling mostly used with datetime data when you have data in one frequency and you want to convery that into another time frequency

```python
import pandas as pd
import numpy as np

# create date range
dates = pd.date_range(start="2024-01-01", periods=800, freq="D")

# create dataframe
df = pd.DataFrame({
    "Date": dates,
    "Sales": np.random.randint(100,500,800)
})

df.set_index("Date", inplace=True)
```

```python
df
```

```python
## week wise sales
df.resample("W").sum()
```

```python
df.resample("W").agg(["sum","mean","max"])
```

```python
## monthly sales
df.resample("M").sum()
```

```python
df.resample("Y").sum()
```
