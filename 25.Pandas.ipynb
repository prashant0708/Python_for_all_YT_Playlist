```python
## function that will take n number of list to create dataframe out of that
import logging

logging.basicConfig(level=logging.INFO)

def list_to_dataframe(*lists): ## 3 list of 5 length  ### case 2 --> 3 list 5,4,5
    try:
        length_iterator = list(map(lambda x : len(x),lists)) # [5,5,5] , ##[5,4,5] -->[5,4]
        
        logging.info(f"length of each list  is length_iterator = {length_iterator}")
        
        x = set(length_iterator) # [4,4,4] --> {4}
        logging.info(f"length of x after removing the duplicate x= {x}")
        
        if len(x)!=1:
            raise ValueError("length of the given list is not same")
        else:
            data=list(zip(*lists))
            logging.info(f"here is your data after zipping{data}")
            result = pd.DataFrame(data)
        return result
    except Exception as e:
        logging.error({"type":"error", "info":str(e)})  
```

```python
## pass three list parameter and convert that into data frame

name= ["Rahul","Sanjeev","Sharma","Avinav"]
age = [20,24,56,45]
salary = [56000, 45000, 340000, 67000]

df4=list_to_dataframe(name,age,salary)
```
