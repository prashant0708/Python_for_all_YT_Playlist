The zip() built in  Python function is used to combine two or more iterables (like lists, tuples, etc.) 
element-wise, 
creating an iterator that aggregates elements from each of the iterables into tuples.

#simple defination 

zip() , pairs elements from two different iterable element wise.

```python
#syntax
zip(iterable1, iterable 2, ....., iterable n)
```

```python
def zips(*iterables):
    iterators = [iter(it) for it in iterables]

    while True:
        result = []
        try:
            for it in iterators:
                result.append(next(it))
            yield tuple(result)
        except StopIteration:
            return   # ✅ stop generator cleanly
```

```python
## adding two list index wise elements

list_1 = [1,2,3,4]

list_2 = [5,6,7,8]

result = list(zip(list_1,list_2))
print(result)

for i , j in result:
    print(i,"+",j ,"----->",i+j)
```

```python
keys = ["name", "age", "city"]
values = ["Gupta", 25, "Delhi"]

result = dict(zip(keys,values))

print(result)

```

```python
## comparing two price and check price increase or descease

old_prices = [100, 200, 300]
new_prices = [110, 190, 310]

price = list(zip(old_prices, new_prices))

for old_price , new_price in price:
    if old_price<new_price:
        print(old_price, new_price, "Price increased")
    else:
        print(old_price, new_price, "Price descreased")
```
