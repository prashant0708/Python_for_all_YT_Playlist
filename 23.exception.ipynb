```python
def read_file(file_path:str):
    try:
        with open(file_path , 'r') as file:
            content = file.readlines()
            return content
    except FileNotFoundError:
        print("File not found")
    except Exception as e:
        print(e)
        
def addition(a,b):
    try:
        return a+b
    except Exception as e:
        print(e)
```

```python
file_path = 123

read_file(file_path)

addition(10,20)
```
