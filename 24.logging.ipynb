What is Logging?
Logging is a way to record messages about a program’s execution for debugging, monitoring, and tracking errors.

Used instead of print() in real applications
Helps understand what happened and when

Simple line:

Logging is like a black box of an airplane — when something crashes, logs tell us what happened before the crash.

# type of logging
1. Debug
2. Info -->information
3. Warning
4. Error
5. Critical

```python
## calculate the area of square

import logging

logging.basicConfig(level=logging.INFO)


def area(edge):
    logging.info("Received the edge as parameter")
    
    result = edge*edge
    logging.info(f"Area of Square calculated {result}")
    
    return result
```

```python
area = area(20)
```

```python
## debug

## checking the varibale value
## function flow

logging.basicConfig(level=logging.WARNING , force =True)
user_input = int(input("Enter your age "))
logging.debug(f"received the user_input{user_input}")
```

```python
## info

## when ever you want to check the normal flow of the code 
logging.info("able to connect to database")
```

```python
## warning
## something is not ideal but still code is running 

logging.warning("Password attempt is getting high") 
```

```python
## error
## when something is wrong and not able to execute the code
logging.error("Failed to connect to db")
```

```python
### critical 
## Application is sutting down 
## system is about to crash

logging.critical("System is crashing")
```

```python
import logging

logging.basicConfig(level=logging.INFO , force=True)

def withdraw(amount, balance):
    if amount>balance:
        logging.warning("you are requesting amount to withdraw more than balance")
    else:
        logging.info("Amont has been released")
        return amount
```

```python
amount= withdraw(2000,1000)
```

```python
def calculator(a,b):
    try:
        logging.info("received paramater a and b")
        if a<b:
            logging.warning("passed b parameter is smaller than a so divison can't be possible")
        else:
            result= a/b
            logging.info("Division is done")
            return result
    except Exception as e:
        logging.error(f"There is error {e}")
```

```python
result = calculator(a=15,b=0)
```
